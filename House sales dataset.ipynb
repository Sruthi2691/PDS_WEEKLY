```python
plt.figure(figsize=(8, 5))
sns.histplot(df["price"], bins=30, kde=True)
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="sqft_living", y="price")
plt.title("Living Area vs House Price")
plt.show()
```

```python
sns.countplot(data=df, x="bedrooms")
plt.title("Number of Bedrooms in Houses")
plt.show()
```

```python
df.groupby("bedrooms")["price"].mean().plot(kind="bar")
plt.title("Average House Price by Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
cols = ["price", "bedrooms", "bathrooms",
        "sqft_living", "sqft_lot"]

sns.heatmap(df[cols].corr(), annot=True, cmap="coolwarm")
plt.title("House Sales Correlation")
plt.show()
```

```python
sns.boxplot(data=df, x="condition", y="price")
plt.title("House Price by Condition")
plt.show()
```

```python
sns.barplot(data=df, x="grade", y="price")
plt.title("Average House Price by Grade")
plt.show()
```

```python
df["date"] = pd.to_datetime(df["date"])

df["house_age"] = (
    df["date"].dt.year - df["yr_built"]
)

print(df[["yr_built", "house_age"]].head())
```

```python
expensive_house = df.loc[df["price"].idxmax()]

print(expensive_house)
```
