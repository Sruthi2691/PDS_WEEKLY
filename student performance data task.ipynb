```python
sns.countplot(data=df, x="gender")
plt.title("Gender Distribution")
plt.show()
```

```python
df[["math score", "reading score", "writing score"]].hist(
    figsize=(10, 5)
)
plt.suptitle("Distribution of Student Scores")
plt.show()
```

```python
df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean().plot(kind="bar")

plt.title("Average Scores by Gender")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```

```python
sns.barplot(
    data=df,
    x="test preparation course",
    y="math score"
)
plt.title("Test Preparation vs Math Score")
plt.show()
```

```python
sns.barplot(
    data=df,
    x="parental level of education",
    y="math score"
)
plt.xticks(rotation=45)
plt.title("Parental Education vs Math Score")
plt.show()
```

```python
scores = df[["math score", "reading score", "writing score"]]

sns.heatmap(scores.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Between Subject Scores")
plt.show()
```

```python
df["total score"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
)

print(df[["total score"]].head())
```

```python
df["average score"] = df[
    ["math score", "reading score", "writing score"]
].mean(axis=1)

print(df[["average score"]].head())
```

```python
highest = df.loc[df["total score"].idxmax()]

print("Highest Total Score:")
print(highest)
```

```python
df.to_csv("student_performance_output.csv", index=False)
print("Dataset saved successfully!")
```
